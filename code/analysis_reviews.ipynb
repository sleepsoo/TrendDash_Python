```python
from transformers import pipeline
import torch

# 감성분석 파이프라인 로드
classifier = pipeline(
    "sentiment-analysis",
    model='sangrimlee/bert-base-multilingual-cased-nsmc'
    )

# 감성 분석 실행
text = "1빠"
result = classifier(text)

print(result)
```

```python
# OpenAI에서 제공하는 GPT 모델을 쉽게 쓸 수 있게 만든 langchain_openai 패키지 활용
from langchain_openai import ChatOpenAI

# 객체 생성
llm = ChatOpenAI(
    temperature=0.4,
    model_name = "gpt-4.1-nano"
)
# https://python.langchain.com/api_reference/openai/chat_models/langchain_openai.chat_models.base.ChatOpenAI.html

review = "다음 글이 긍정적인지 중립적인지 부정적인지 단답으로 말해주세요.(선택지는 오직 긍정/중립/부정 중 하나)"

print(f"[답변] : {llm.invoke(review)}")

```
