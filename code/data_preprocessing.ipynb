## 7가지 감정 데이터

```python
import pandas as pd

df_exel = pd.read_excel("./dataset_raw/한국어_단발성_대화_데이터셋.xlsx")
emotions = ['행복', '분노', '혐오', '중립']
df_exel = df_exel[df_exel['Emotion'].isin(emotions)]
df_exel.columns = ['내용', 'label']
df_exel = df_exel[['label', '내용']]
df_exel
```

```python
label_map = {'행복':'긍정', '분노':'부정', '혐오':'부정'}
df_exel['label'] = df_exel['label'].map(label_map).fillna(df_exel['label'])
df_exel
```

```python
df_exel.to_csv()
```

## 네이버 평점 데이터

```python
df_txt = pd.read_csv("./dataset_raw/naver_shopping.txt", sep='\t', header=None, names = ['label', '내용'])
df_txt
```

```python
label_map = {1:"부정", 2:"부정", 4:"긍정", 5:"긍정"}
df_txt['label'] = df_txt["label"].map(label_map).fillna(df_txt['label'])
df_txt
```

## steam 리뷰 데이터

```python
df_txt2 = pd.read_csv("./dataset_raw/steam.txt", sep='\t', header=None, names = ['label', '내용'])
df_txt2
```

```python
label_map = {0:"부정", 1:"긍정"}
df_txt2['label'] = df_txt2['label'].map(label_map).fillna(df_txt2['label'])
df_txt2
```

## 내용 합치기

```python
df_concat = pd.concat([df_txt,df_txt2, df_exel], axis=0)
df_concat
```

```python
df_concat.to_csv('review_txt', sep='\t', index=False, encoding='utf-8')
```
